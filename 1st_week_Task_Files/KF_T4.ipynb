```python
from transformers import pipeline

chatbot = pipeline("text-generation", model="facebook/opt-350m")

def health_assistant(user_query):

    system_prompt = (
        "You are a friendly and helpful medical assistant. "
        "Provide general information but always include a safety disclaimer. "
        "User says: "
    )

    full_prompt = system_prompt + user_query

    response = chatbot(full_prompt, max_length=150, do_sample=True, temperature=0.7)

    return response[0]['generated_text'].replace(system_prompt, "")

print("Response 1:", health_assistant("What causes a sore throat?"))
print("\nResponse 2:", health_assistant("Is paracetamol safe for children?"))
```

```python
from google.colab import files
import pandas as pd
from transformers import pipeline

# 1. Model Setup
chatbot = pipeline("text-generation", model="facebook/opt-350m")

# 2. Data save karne ke liye ek khali list banayein
conversation_history = []

def health_assistant(user_query):
    system_prompt = (
        "You are a friendly and helpful medical assistant. "
        "Provide general information but always include a safety disclaimer. "
        "User says: "
    )

    full_prompt = system_prompt + user_query
    response = chatbot(full_prompt, max_length=150, do_sample=True, temperature=0.7)
    bot_reply = response[0]['generated_text'].replace(system_prompt, "")

    # 3. Har query aur reply ko list mein save karein
    conversation_history.append({
        "User_Query": user_query,
        "Bot_Response": bot_reply
    })

    return bot_reply

# Test Queries (Aap jitni chahein queries add kar sakte hain)
print("Response 1:", health_assistant("What causes a sore throat?"))
print("\nResponse 2:", health_assistant("Is paracetamol safe for children?"))

# =========================================================
# 4. POWER BI KE LIE DATA DOWNLOAD KARNE KI COMMANDS
# =========================================================

# List ko tabular format (DataFrame) mein convert karein
df = pd.DataFrame(conversation_history)

# CSV file banayein
df.to_csv('chatbot_interactions.csv', index=False)

# File download karein
files.download('chatbot_interactions.csv')
```
